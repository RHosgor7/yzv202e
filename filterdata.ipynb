```python
import pandas as pd

for i in range(1, 11):
    input_file = f'nasa_block_{i}.csv'
    output_file = f'nasa_annual_summary{i}.csv'
    
    # Load the data
    df = pd.read_csv(input_file)
    
    # Filter rows where date ends with '13'
    annual_summary = df[df['date'].astype(str).str.endswith('13')]
    
    # Save to new CSV file
    annual_summary.to_csv(output_file, index=False)
    
    print(f"Saved annual summary rows to '{output_file}'")
```

```python
import pandas as pd

for i in range(1, 11):
    input_file = f'nasa_annual_summary{i}.csv'
    output_file = f'nasa_annual_summary_averaged{i}.csv'
    
    # Load the filtered annual summary file
    df = pd.read_csv(input_file)
    
    # Group by (lat, lon) and calculate mean for each parameter
    averaged = df.groupby(['lat', 'lon'], as_index=False).agg({
        'ALLSKY_SFC_SW_DWN': 'mean',
        'ALLSKY_KT': 'mean',
        'CLOUD_AMT': 'mean',
        'PRECTOTCORR': 'mean'
    })
    
    # Round the results
    averaged = averaged.round({
        'ALLSKY_SFC_SW_DWN': 4,
        'ALLSKY_KT': 4,
        'CLOUD_AMT': 4,
        'PRECTOTCORR': 4
    })
    
    # Save to new CSV file
    averaged.to_csv(output_file, index=False)
    
    print(f"Saved averaged summary per point (rounded) to '{output_file}'")
```

```python
import pandas as pd

# List to hold all DataFrames
dfs = []

for i in range(1, 11):
    input_file = f'nasa_annual_summary_averaged{i}.csv'
    
    # Read each file and append to list
    df = pd.read_csv(input_file)
    dfs.append(df)

# Concatenate all DataFrames
combined_df = pd.concat(dfs, ignore_index=True)

# Save combined file
combined_df.to_csv('nasa_annual_summary_averaged_all.csv', index=False)

print("Saved combined averaged summary to 'nasa_annual_summary_averaged_all.csv'")
```
