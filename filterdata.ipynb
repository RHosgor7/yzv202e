```python
import pandas as pd

# List to hold all DataFrames
dfs = []

for i in range(1, 11):
    input_file = f'data/nasa_block_{i}.csv' 
    
    # Read each file and append to list
    df = pd.read_csv(input_file)
    dfs.append(df)

# Concatenate all DataFrames
combined_df = pd.concat(dfs, ignore_index=True)

# Save combined file
combined_df.to_csv('nasa_all_data', index=False)

print("Saved combined averaged summary to 'nasa_all_data.csv'")
```
